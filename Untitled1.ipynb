```python
import pandas as pd
import numpy as np
data = pd.read_csv("World Happiness Report.csv")
import matplotlib.pyplot as plt

# data.loc[(data["Calories"]>375.0),"Health"] = "Unhealthy"
# data.loc[((data["Calories"]<=375.0) & (data["Calories"]>180.0)),"Health"] = "Average"
# data.loc[(data["Calories"]<=180.0),"Health"] = "healthy"
# data =data.groupby("Health").agg({"Calories" : "count"})
# print(data)
for i in range(1,5):
    i=i*20
    df = data.head(i)
    df=df.groupby("Region").agg({"Region":"count"})
    c=["b","g","y","r"]
    i=i//20
    df.plot(kind="bar",color=c[i-1])
    plt.grid(axis="y", linestyle="--", alpha=0.4)
    plt.xticks(rotation=45)
```
